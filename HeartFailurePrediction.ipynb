#Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#Importing Dataset

```python
dataset = pd.read_csv('heart.csv')
x = dataset.iloc[:,:-1].values
y = dataset.iloc[:,-1].values
```

```python
print(x)
```

```python
print(y)
```

#Data Preprocessing

#Label Encoding


```python
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
le2 = LabelEncoder()
le6 = LabelEncoder()
le8 = LabelEncoder()
le10 = LabelEncoder()
x[:,1] = le1.fit_transform(x[:,1])
x[:,2] = le2.fit_transform(x[:,2])
x[:,6] = le6.fit_transform(x[:,6])
x[:,8] = le8.fit_transform(x[:,8])
x[:,10] = le10.fit_transform(x[:,10])
```

```python
print(x)
```

#Splitting Dataset into Training set and Test set

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
print(X_train)
```

```python
print(Y_test)
```

#Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
print(X_test)
```

#Training Dataset

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
model_logistic = LogisticRegression()
model_logistic.fit(X_train,Y_train)
```

```python
#SVM
from sklearn.svm import SVC
model_svm = SVC()
model_svm.fit(X_train,Y_train)
```

```python
#KNeighbours
from sklearn.neighbors import KNeighborsClassifier
model_kneighbors = KNeighborsClassifier(n_neighbors=8)
model_kneighbors.fit(X_train,Y_train)
```

```python
#Decision Tree
from sklearn.tree import DecisionTreeClassifier
model_decision = DecisionTreeClassifier()
model_decision.fit(X_train,Y_train)
```

```python
#Random Forest
from sklearn.ensemble import RandomForestClassifier
model_randomforest = RandomForestClassifier(n_estimators=200)
model_randomforest.fit(X_train,Y_train)
```

```python
#XGBoost
from xgboost import XGBClassifier
model_xgb = XGBClassifier()
model_xgb.fit(X_train,Y_train)
```

```python
y_pred_logistic = model_logistic.predict(X_test)
y_pred_svm = model_svm.predict(X_test)
y_pred_kneighbors = model_kneighbors.predict(X_test)
y_pred_decision = model_decision.predict(X_test)
y_pred_randomforest = model_randomforest.predict(X_test)
y_pred_xg = model_xgb.predict(X_test)
```

#Testing Accuracy on different ML Algorithms

```python
from sklearn.metrics import accuracy_score
Logistic_Acc = accuracy_score(Y_test,y_pred_logistic)
SVM_Acc = accuracy_score(Y_test,y_pred_svm)
KNeighbors_Acc = accuracy_score(Y_test,y_pred_kneighbors)
DecTree_Acc = accuracy_score(Y_test, y_pred_decision)
Random_Acc = accuracy_score(Y_test,y_pred_randomforest)
Xg_Acc = accuracy_score(Y_test,y_pred_xg)
```

```python
print(SVM_Acc)
print(Logistic_Acc)
print(KNeighbors_Acc)
print(DecTree_Acc)
print(Random_Acc)
print(Xg_Acc)
```

#Plotting Accuracy of different ML Algorithms

```python
plt.figure(figsize=(12, 6))
plt.bar("Logistic Regression",Logistic_Acc,width=0.6)
plt.bar("KNeighbors",KNeighbors_Acc,width=0.6)
plt.bar("Support Vector Machine",SVM_Acc,width=0.6)
plt.bar("Decision Tree",DecTree_Acc,width=0.6)
plt.bar("Random Forest",Random_Acc,width=0.6)
plt.bar("XGBoost",Xg_Acc,width=0.6)
plt.xlabel("Machine Learning Algorithm")
plt.ylabel("Accuracy")
```
